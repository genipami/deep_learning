#### Task 4: Python vs NumPy

Use pure Python and `numpy` to create a Python list/`numpy`:

- array of $10$ zeros;
- array of $10$ ones;
- array of $10$ fives;
- array of $10$ zeroes, the fifth value of which is $1$;
- array of all the integers from $10$ to $50$;
- array of all the even integers from $10$ to $50$;
- array that is the reversal of the list `[1, 2, 3]` (save these values in an object and use both methods to reverse that object);
- a 3x3 matrix with values ranging from $0$ to $8$;
- a 3x3 identity matrix;
- a 5x5 matrix with values 1,2,3,4 just below the diagonal.

````python
print("array of 10 zeros")
print("Python:", [0] * 10)
print("Numpy:", np.zeros(10))
print()

print("array of 10 ones")
print("Python:", [1] * 10)
print("Numpy:", np.ones(10))
print()

print("array of 10 fives")
print("Python:", [5] * 10)
print("Numpy:", np.full(10, 5.))
print()

print("array of 10 zeroes, the fifth value of which is 1")
print("Python:", [0] * 4 + [1] + [0] * 5)
fifth_one = np.zeros(10)
fifth_one[4] = 1
print("Numpy:", fifth_one)
print()

print("array of all the integers from 10 to 50")
ten_to_fifty = []
for i in range(10, 51):
    ten_to_fifty.append(i)
print("Python:", ten_to_fifty)
print("Numpy:", np.arange(10, 51, 1))
print()

print("array of all the even integers from 10 to 5")
ten_to_fifty_step_two = []
for i in range(10, 51, 2):
    ten_to_fifty_step_two.append(i)
print("Python:", ten_to_fifty_step_two)
print("Numpy:", np.arange(10, 51, 2))
print()

print("array that is the reversal of the list `[1, 2, 3]`")
arr = [1, 2, 3]
np_arr = np.array(arr)
print("Python:", arr[::-1])
print("Numpy:", np_arr[::-1])
print()

print("a 3x3 matrix with values ranging from 0 to 8")
arr = [[0 + a * 3, 1 + a * 3, 2 + a * 3] for a in range(3)]
np_arr = np.arange(0,9)
print("Python:", arr)
print("Numpy:", np_arr.reshape(3,3))
print()

print("a 3x3 identity matrix")
arr = []
for i in range(3):
    temp = []
    for j in range(3):
        temp.append(int(i == j))
    arr.append(temp)
print("Python:", arr)
print("Numpy:", np.eye(3))
print()

print("a 5x5 matrix with values 1,2,3,4 just below the diagonal")
arr = []
for i in range(5):
    temp = []
    for j in range(5):
        temp.append(i if i - 1 == j else 0)
    arr.append(temp)

np_arr = np.eye(5, 5, k=-1)
for i in range(5):
    np_arr[i] = np_arr[i] * i

print("Python:", arr)
print("Numpy:", np_arr)
print()
````

#### Task 6: Use NumPy to generate:

- a random number between 0 and 1;
- an array of 25 random numbers sampled from a standard normal distribution;
- a 3x3x3 array with random values drawn from the uniform distribution;
- a 10x10 array with random values drawn from the uniform distribution and find the minimum and maximum values;
- a random array with values drawn from the uniform distribution of size 30 and find the mean value;
- and multiply a 5x3 matrix by a 3x2 matrix.

**Note**: Your values should be different from the one shown below due to randomness.

````python
print("a random number between 0 and 1")
print(np.random.rand(1)[0])
print()

print("an array of 25 random numbers sampled from a standard normal distribution")
print(np.random.normal(loc=0, scale=1.0, size=25))
print()

print("a 3x3x3 array with random values drawn from the uniform distribution")
print(np.random.uniform(low=0.0, high=1.0, size=27).reshape((3,3,3)))
print()

print("a 10x10 array with random values drawn from the uniform distribution and find the minimum and maximum values")
print("The array:")
array = np.random.uniform(low=0.0, high=1.0, size=100).reshape((10,10))
print(array)
print("Min and max values:", array.min(), array.max())
print()

print("a random array with values drawn from the uniform distribution of size 30 and find the mean value")
print("The array:")
array = np.random.uniform(low=0.0, high=1.0, size=30)
print(array)
print("It's mean:", array.mean())
print()

print("multiplying a 5x3 matrix by a 3x2 matrix each with random values")
first = np.random.uniform(low=0.0, high=1.0, size=15).reshape((5,3))
second = np.random.uniform(low=0.0, high=1.0, size=6).reshape((3,2))
print(np.matmul(first, second))
````
